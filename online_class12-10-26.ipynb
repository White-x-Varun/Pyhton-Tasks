```python

courses = {
    "Python": {"instructor": "Rahul", "duration": 60},
    "Java": {"instructor": "Priya", "duration": 45},
    "SQL": {"instructor": "Soniya", "duration": 45}
}

for course, details in courses.items():
    if course == "Python":
        print("Instructor:", details["instructor"])
        print("Duration:", details["duration"], "hrs")
```

```python
sales=[
    ["Aman",12000,15000,18000],
    ["Priya",20000,17000,14000],
    ["Rohit",16000,22000,19000]
]
highest_sale=0
top_salesperson=""

for person in sales:
    name=person[0]
    total=sum(person[1:])

    print("Name",name)
    print("Total",total)

    if total>highest_sale:
        highest_sale=total
        top_salesperson=name

print("top_salesperson",top_salesperson)
print("highest_sale",highest_sale)
    
    
```
