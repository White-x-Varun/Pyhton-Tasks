```python
numbers = [20, 55, 30, 80, 45, 90, 10]

a= [n for n in numbers if n > 50]

print(a)
```

```python
numbers = [10, 25, 35, 60, 45, 15, 50, 70]

new_list = [n for n in numbers if 20 <= n <= 50]

print(new_list)
```

```python
numbers = [10, 15, 20, 30, 45, 50, 60, 75]

new_list = [n for n in numbers if n % 3 == 0 and n % 5 == 0]

print(new_list)
```

```python
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]

new_list = [n for n in names if len(n) > 4]

print(new_list)
```

```python
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]

new_list = [name for name in names if name.startswith("R")]

print(new_list)
```

```python
text = "programming"

new_list = [n for n in text if n in "aeiou"]

print(new_list)
```

```python
data = "Python123Programming456"

new_list = [n for n in data if n.isdigit()]

print(new_list)
```

```python
numbers = [10, 15, 20, 25, 30]

new_list = ["Even" if n % 2 == 0 else "Odd" for n in numbers]

print(new_list)
```

```python
numbers = [10, -5, 20, -8, 30, -2]

new_list = [0 if n < 0 else n for n in numbers]

print(new_list)
```

```python
prices = [500, 1200, 800, 2000, 1500]

new_list = [n * 0.9 if n > 1000 else n * 0.95 for n in prices]

print(new_list)
```

```python
employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

new_list = [
    emp["name"] for emp in employees
    if emp["salary"] > 50000
]

print(new_list)
```
